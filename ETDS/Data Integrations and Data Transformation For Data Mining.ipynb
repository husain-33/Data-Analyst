### Practical: Data Integrations and Data Transformation For Data Mining

```python
import pandas as pd 
import numpy as np 
```

```python
data1_hus= {
    'ID':[1,2,3,4],
    'Name':['Husain','Sonal','Zain','David'],
    'Age':[25,18,35,49]
    
}

data2_hus= {
    'ID':[3,4,5,6],
    'Gender':['M','F','M','M'],
    'Salary':[80000,82000,94000,55000]
    
}
df1_hus= pd.DataFrame(data1_hus)
df2_hus=pd.DataFrame(data2_hus)
print(df1_hus.head())
print("\n\n",df2_hus.head())
```

### Data Integration

```python
#Tight Coupling(Join DAtasets on a comman Key)

tight_hus=pd.merge(df1_hus,df2_hus,on='ID',how='inner')
print("Tight Coupling Result:\n",tight_hus)
```

```python
#Loose Coupling(Concatenate datasets)

loose_hus= pd.concat([df1_hus.set_index('ID'),df2_hus.set_index('ID')], axis=1).reset_index()
print("Loose Coupling Result\n:",loose_hus)
```

```python
loose_hus['Smooth_Age']= loose_hus['Age'].rolling(window=2, min_periods=1).mean()
print("Smmothing:\n",loose_hus[['ID','Age','Smooth_Age']])
```

```python
agg=loose_hus.groupby('Gender')['Salary'].sum().reset_index()
print("Aggregations:\n",agg)
```

```python
bins= [0,20,30,40,50]
lab=['Teen','Young','Adult','Senior']
loose_hus["Age_Group"]=pd.cut(loose_hus["Age"], bins=bins, labels=lab)
print("Discretization:\n", loose_hus[['ID','Age','Age_Group']])
```

```python
loose_hus['Age_Salary_ratio']=loose_hus['Age'] / loose_hus['Salary']
print('\nAttribute Contribution:\n',loose_hus[['ID','Age','Salary','Age_Salary_ratio']])
```

```python
salary_bins=[0,60000,80000,100000]
sal_lab=['low','Medium','high']
loose_hus['Salary_range']= pd.cut(loose_hus['Salary'],bins=salary_bins, labels=sal_lab)
print("\nGeneralization:\n",loose_hus[['ID','Salary','Salary_range']])
```

```python
loose_hus['Age_MinMax']= (loose_hus['Age'] - loose_hus['Age'].min()) /(loose_hus['Age'].max() - loose_hus['Age'].min())
print("Min-Max Normalization:\n",loose_hus[['ID','Age','Age_MinMax']])
```

```python
loose_hus['Age_Z']= (loose_hus['Age'] -loose_hus['Age'].mean()) / loose_hus['Age'].std()
print("\n Z-Score Normalization:\n",loose_hus[['ID','Age','Age_Z']])
```

```python
scal_fact= 10 ** np.ceil(np.log10(loose_hus['Age'].abs().max()))
loose_hus['Age_DecimalScaling'] = loose_hus['Age'] / scal_fact
print("\n Decimal Scaling:\n",loose_hus[['ID','Age','Age_DecimalScaling']])
```
